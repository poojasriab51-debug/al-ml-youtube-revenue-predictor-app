# YouTube Ad Revenue Analysis
This notebook loads `youtube_ad_revenue_dataset (2) copy.csv` and runs a few basic exploratory data analysis steps.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)

path = 'youtube_ad_revenue_dataset (2) copy.csv'
df = pd.read_csv(path)
df.head()
```

```python
df.shape, df.dtypes
```

```python
df.isna().sum().sort_values(ascending=False).head(20)
```

```python
num_cols = df.select_dtypes(include='number').columns.tolist()
num_cols
```

```python
if num_cols:
    df[num_cols].describe().T
else:
    print('No numeric columns found.')
```

```python
# Plot distributions for numeric columns (top 6 by variance)
if num_cols:
    top = (df[num_cols].var().sort_values(ascending=False).head(6).index.tolist())
    plt.figure(figsize=(14, 8))
    for i, c in enumerate(top, 1):
        plt.subplot(2, 3, i)
        sns.histplot(df[c].dropna(), kde=True)
        plt.title(c)
    plt.tight_layout()
    plt.show()
else:
    print('No numeric columns found.')
```

```python
# Correlation heatmap for numeric columns
if len(num_cols) >= 2:
    plt.figure(figsize=(10, 8))
    sns.heatmap(df[num_cols].corr(numeric_only=True), annot=False, cmap='coolwarm')
    plt.title('Correlation Heatmap (Numeric Columns)')
    plt.show()
else:
    print('Not enough numeric columns for correlation heatmap.')
```
